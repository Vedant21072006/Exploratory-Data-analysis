```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('titanic')
df.head()
```

```python
# Find the categorical and numerical data
num_cols=df.select_dtypes(include=['number']).columns
cat_cols=df.select_dtypes(exclude=['number']).columns
print("Numerical columns: ",num_cols)
print("Categorical columns: ",cat_cols)
```

# Categorical data analysis

```python
# Counts:
# 1. Number of males and females
sns.catplot(x='sex',kind='count',data=df)
plt.title('Number of males and females')
plt.show()
# 2. Number of peoples in each class
sns.catplot(x='pclass',kind='count',data=df)
plt.title('Number of peoples in each class')
plt.show()
# 3.Number of adult males
sns.catplot(x='adult_male',kind='count',data=df)
plt.title('Number of adult and non-adult males')
plt.show()
# 4. People from boarded from each port
sns.catplot(x='embark_town',kind='count',data=df)
plt.title('Number of people that boarded from each port')
plt.show()
# 5. No. of people that are alone
sns.catplot(x='alone',kind='count',data=df)
plt.title('Number of people who were alone')
plt.show()
# 6. Number of people who surived
sns.catplot(x='survived',kind='count',data=df)
plt.title('Number of people who survived')
plt.show()
```

```python

# Number of people survived on basis of gender
sns.catplot(x='survived',kind='count',data=df,hue='sex')
plt.title('Survival on basis on gender')
plt.show()
# Number of people survived on basis of adult male
sns.catplot(x='survived',kind='count',data=df,hue='adult_male')
plt.title('Survival on basis of adult_male')
plt.show()
# Number of people survived on basis they was alone or not
sns.catplot(x='survived',kind='count',data=df,hue='alone')
plt.title('Survival on basis wheather they were alone or not')
plt.show()
# Survival on the basis of which port they boarded
sns.catplot(x='survived',kind='count',data=df,hue='embark_town')
plt.title('Survival on basis on which port they boarded from')
plt.show()
```

# Numerical data analysis

```python
# 1.Age analysis
sns.displot(x=df['age'],kde=True,bins=50)
plt.title('Distplot of ages')
plt.show()
# age analysis with boxplot
sns.boxplot(x=df['age'])
plt.title('Box plot of ages')
plt.show()
```

```python
# analysis on fares
sns.displot(df['fare'],bins=20,kde=True)
plt.title('Distplot on fares')
plt.show()
sns.boxplot(df['fare'])
```

```python
# skewness
df[['age','fare']].skew()
```

```python
# median and mean
df['age'].mean()
```
